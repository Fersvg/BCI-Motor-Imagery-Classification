```python
import pandas as pd
import numpy as np
import scipy.io
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from mne.decoding import CSP
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
from sklearn.decomposition import PCA

# Load data
X1 = pd.read_csv('features_model1.csv') 
X2 = pd.read_csv('features_model2.csv')
data_csp = scipy.io.loadmat('trials_CSP.mat')

X_3ch = np.transpose(data_csp['X_trials_3ch'], (2,0,1))
X_7ch = np.transpose(data_csp['X_trials_7ch'], (2,0,1))

X_3ch_mu = np.transpose(data_csp['X_trials_3ch_mu'], (2,0,1))
X_3ch_beta = np.transpose(data_csp['X_trials_3ch_beta'], (2,0,1))

X_7ch_mu = np.transpose(data_csp['X_trials_7ch_mu'], (2,0,1))
X_7ch_beta = np.transpose(data_csp['X_trials_7ch_beta'], (2,0,1))

y_csp = data_csp['y_trials'].ravel()

# Model 1 and Model 2 Features

y = X1['label'].values
X1 = X1.drop(columns=['label']).values
X2 = X2.drop(columns=['label']).values

le = LabelEncoder()
y_enc = le.fit_transform(y)  # left=0, rest=1, right=2

# Global Train Test Split
X1_train, X1_test, y_train, y_test = train_test_split(
    X1, y_enc, test_size=0.2, stratify=y_enc, random_state=42
)

X2_train, X2_test, _, _ = train_test_split(
    X2, y_enc, test_size=0.2, stratify=y_enc, random_state=42
)

# Normalize Model 1 and Model 2 features
scaler1 = StandardScaler()
X1_train = scaler1.fit_transform(X1_train)
X1_test  = scaler1.transform(X1_test)

scaler2 = StandardScaler()
X2_train = scaler2.fit_transform(X2_train)
X2_test  = scaler2.transform(X2_test)

# MODEL 1 (ERD(Cz, C2, C4) and MRCP(Cz)
print("\n MODEL 1")

models1 = {
    "LDA": LinearDiscriminantAnalysis(solver='lsqr', shrinkage='auto'),
    "SVM": SVC(kernel='rbf', C=1, gamma=0.1, class_weight='balanced'),
    "LogReg": LogisticRegression(max_iter=1000, class_weight='balanced'),
    "RF": RandomForestClassifier(n_estimators=100),
}

best_acc_m1 = 0
best_name_m1 = ""

for name_model, model in models1.items():

    model.fit(X1_train, y_train)
    y_pred = model.predict(X1_test)

    acc = accuracy_score(y_test, y_pred)
    print(f"{name_model} Accuracy:", acc)

    ConfusionMatrixDisplay.from_predictions(
        y_test, y_pred, display_labels=le.classes_
    )
    plt.title(f"Model 1 - {name_model}")
    plt.show()

    if acc > best_acc_m1:
        best_acc_m1 = acc
        best_name_m1 = name_model

print(f"\nBEST MODEL 1: {best_name_m1} : {best_acc_m1}")

# MODEL 2 (ERD(Cz, C2, C4) and MRCP(Cz, C3, C4))
print("\n MODEL 2")

models2 = {
    "LDA": LinearDiscriminantAnalysis(solver='lsqr', shrinkage='auto'),

    "LDA + PCA": Pipeline([
        ('pca', PCA(n_components=0.95)),
        ('lda', LinearDiscriminantAnalysis(solver='lsqr', shrinkage='auto'))
    ]),

    "SVM": SVC(kernel='rbf', C=1, gamma=0.1, class_weight='balanced'),

    "SVM + PCA": Pipeline([
        ('pca', PCA(n_components=0.95)),
        ('svm', SVC(kernel='rbf', C=1, gamma=0.1, class_weight='balanced'))
    ]),

    "LogReg": LogisticRegression(max_iter=1000, class_weight='balanced'),

    "LogReg + PCA": Pipeline([
        ('pca', PCA(n_components=0.95)),
        ('logreg', LogisticRegression(max_iter=1000, class_weight='balanced'))
    ]),

    "RF": RandomForestClassifier(n_estimators=100),

    "RF + PCA": Pipeline([
        ('pca', PCA(n_components=0.95)),
        ("RF", RandomForestClassifier(n_estimators=100))
    ])
}

best_acc_m2 = 0
best_name_m2 = ""

for name_model, model in models2.items():

    model.fit(X2_train, y_train)
    y_pred = model.predict(X2_test)

    acc = accuracy_score(y_test, y_pred)
    print(f"{name_model} Accuracy:", acc)

    ConfusionMatrixDisplay.from_predictions(
        y_test, y_pred, display_labels=le.classes_
    )
    plt.title(f"Model 2 - {name_model}")
    plt.show()

    if acc > best_acc_m2:
        best_acc_m2 = acc
        best_name_m2 = name_model

print(f"\nBEST MODEL 2: {best_name_m2} : {best_acc_m2}")

# CSP Models
def run_csp_pipeline(X, y, name="MODEL"):

    print(f"\n {name}")

    # Normalization
    std = np.std(X, axis=2, keepdims=True)
    std[std == 0] = 1
    X = X / std
    # CSP 1: REST vs MOVEMENT
    print("\n CSP1 (Rest vs Movement)")

    # CSP1 Binary labels
    y_csp1 = (y != 0).astype(int)

    # CSP1 train test split
    X_train, X_test, y_train_csp1, y_test_csp1 = train_test_split(
    X, y_csp1, test_size=0.2, stratify=y_csp1, random_state=42
    )
    print("X shape:", X_train.shape)
    print("y:", np.unique(y_train_csp1, return_counts=True))

    models_csp1 = {
        "LDA": Pipeline([
            ('csp', CSP(n_components=8, reg='ledoit_wolf', log=False)),
            ('lda', LinearDiscriminantAnalysis(solver='lsqr', shrinkage='auto'))
        ]),
        "SVM": Pipeline([
            ('csp', CSP(n_components=8, reg='oas', log=False)),
            ('scaler', StandardScaler()),
            ('svm', SVC(kernel='rbf', C=10, gamma=0.1, class_weight='balanced'))
        ]),
        "LogReg": Pipeline([
            ('csp', CSP(n_components=8, reg='oas', log=False)),
            ('scaler', StandardScaler()),
            ('logreg', LogisticRegression(max_iter=1000, class_weight='balanced'))
        ]),
        "RF": Pipeline([
            ('csp', CSP(n_components=8, reg='oas', log=False)),
            ('rf', RandomForestClassifier(n_estimators=100))
        ])       
    }

    best_acc_csp1 = 0
    best_model_csp1 = None
    best_name_csp1 = ""

    for name_model, model in models_csp1.items():

        model.fit(X_train, y_train_csp1)
        y_pred = model.predict(X_test)

        acc = accuracy_score(y_test_csp1, y_pred)
        print(f"{name_model} CSP1 Accuracy:", acc)

        ConfusionMatrixDisplay.from_predictions(
            y_test_csp1, y_pred,
            display_labels=['rest','movement']
        )
        plt.title(f"{name} - CSP1 Confusion Matrix - {name_model}")
        plt.show()

        if acc > best_acc_csp1:
            best_acc_csp1 = acc
            best_model_csp1 = model
            best_name_csp1 = name_model

    print(f"Best CSP1: {best_name_csp1}, Accuarcy: {best_acc_csp1}")
    
    # CSP 2: LEFT vs RIGHT
    print("\n CSP2 (Left vs Right)")

    # left vs right mask
    mask = y != 0

    X_lr = X[mask]
    y_lr = y[mask]

    # left = 0, right = 1
    y_lr = (y_lr == 1).astype(int)

    # CSP 2 train test split
    X_train_csp2, X_test_csp2, y_train_csp2, y_test_csp2 = train_test_split(
    X_lr, y_lr, test_size=0.2, stratify=y_lr, random_state=42
    )

    models_csp2 = {
        "LDA": Pipeline([
            ('csp', CSP(n_components=8, reg='ledoit_wolf', log=False)),
            ('lda', LinearDiscriminantAnalysis(solver='lsqr', shrinkage='auto'))
        ]),
        "SVM": Pipeline([
            ('csp', CSP(n_components=8, reg='oas', log=False)),
            ('scaler', StandardScaler()),
            ('svm', SVC(kernel='rbf', C=10, gamma=0.1, class_weight='balanced'))
        ]),
        "LogReg": Pipeline([
            ('csp', CSP(n_components=8, reg='oas', log=False)),
            ('scaler', StandardScaler()),
            ('logreg', LogisticRegression(max_iter=1000, class_weight='balanced'))
        ]),
        "RF": Pipeline([
            ('csp', CSP(n_components=8, reg='oas', log=False)),
            ('rf', RandomForestClassifier(n_estimators=100))
        ])      
    }

    best_acc_csp2 = 0
    best_model_csp2 = None
    best_name_csp2 = ""

    for name_model, model in models_csp2.items():

        model.fit(X_train_csp2, y_train_csp2)
        y_pred = model.predict(X_test_csp2)

        acc = accuracy_score(y_test_csp2, y_pred)
        print(f"{name_model} CSP2 Accuracy: ", acc)

        ConfusionMatrixDisplay.from_predictions(
            y_test_csp2, y_pred,
            display_labels=['left','right']
        )
        plt.title(f"{name} - CSP2 Confusion Matrix - {name_model}")
        plt.show()

        if acc > best_acc_csp2:
            best_acc_csp2 = acc
            best_model_csp2 = model
            best_name_csp2 = name_model

    print(f" Best CSP2: {best_name_csp2}, Accuarcy: {best_acc_csp2}")

    # CASCADE (BEST MODELS)
    print("\n CASCADE")

    # Para cascade usamos TODO el dataset original split
    X_train_full, X_test_full, y_train_full, y_test_full = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
    )

    y_pred_final = np.zeros(len(y_test_full))

    for i in range(len(X_test_full)):

        x_trial = X_test_full[i].reshape(1, X_test_full.shape[1], X_test_full.shape[2])

        pred1 = best_model_csp1.predict(x_trial)[0]

        if pred1 == 0:
            y_pred_final[i] = 0
        else:
            pred2 = best_model_csp2.predict(x_trial)[0]
            y_pred_final[i] = -1 if pred2 == 0 else 1

    acc_final = accuracy_score(y_test_full, y_pred_final)

    print(f" Cascade (CSP1: {best_name_csp1} + CSP2: {best_name_csp2}), Accuracy:", acc_final)

    ConfusionMatrixDisplay.from_predictions(
        y_test_full, y_pred_final,
        display_labels=['left','rest','right']
    )
    plt.title(f"{name} - Confusion Matrix - Cascade")
    plt.show()

    return acc_final

results = {}

results["Model 1"] = best_acc_m1

results["Model 2"] = best_acc_m2

results["3ch"] = run_csp_pipeline(X_3ch, y_csp, "3ch BASE")
results["7ch"] = run_csp_pipeline(X_7ch, y_csp, "7ch BASE")

results["3ch_mu"] = run_csp_pipeline(X_3ch_mu, y_csp, "3ch MU")
results["3ch_beta"] = run_csp_pipeline(X_3ch_beta, y_csp, "3ch BETA")

results["7ch_mu"] = run_csp_pipeline(X_7ch_mu, y_csp, "7ch MU")
results["7ch_beta"] = run_csp_pipeline(X_7ch_beta, y_csp, "7ch BETA")

X_3ch_mubeta = np.concatenate([X_3ch_mu, X_3ch_beta], axis=1)
X_7ch_mubeta = np.concatenate([X_7ch_mu, X_7ch_beta], axis=1)

results["3ch_mubeta"] = run_csp_pipeline(X_3ch_mubeta, y_csp, "3ch MU+BETA")
results["7ch_mubeta"] = run_csp_pipeline(X_7ch_mubeta, y_csp, "7ch MU+BETA")

plt.figure(figsize=(10,5))
plt.bar(results.keys(), results.values())
plt.xticks(rotation=45)
plt.ylabel("Accuracy")
plt.title("Models Comparison")
plt.grid()
plt.show()
```
